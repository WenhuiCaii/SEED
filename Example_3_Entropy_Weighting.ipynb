# Example 3: Simple Entropy Weighting

```python
# Import Packages
import pandas as pd
import numpy as np
```

```python
# Import Data, here we will using a sample data to demo
data = {
    "SA1_CODE21": ["A", "B", "C", "D"],
    "Indicator_1": [0.2, 0.3, np.nan, 0.5],
    "Indicator_2": [0.5, 0.3, 0.2, np.nan]
}
df = pd.DataFrame(data)
```

```python
# Extract indicator columns for further calculation
X = df.drop(columns=["SA1_CODE21"]).values.astype(float)
n, m = X.shape
```

```python
# Calculate weights
col_sums = np.nansum(X, axis=0)
p = X / col_sums
p_safe = np.where(p > 0, p, np.nan)
k = 1.0 / np.log(np.sum(~np.isnan(X), axis=0))
entropy = -k * np.nansum(p_safe * np.log(p_safe), axis=0)
d = 1 - entropy
w = d / np.nansum(d)
weights = pd.Series(w, index=df.columns[1:], name="weight")
print("Entropy weights:")
print(weights)
```

```python
# Composite Index
composite = np.nansum(X * w, axis=1)
df["Composite_Index"] = composite
print("\nFinal composite index:")
print(df)
```
